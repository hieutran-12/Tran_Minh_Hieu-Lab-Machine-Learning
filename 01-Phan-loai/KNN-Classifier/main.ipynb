```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt   
 
```

```python
df = pd.read_csv(r'D:\Hoc-May\Lam_Bai_Lab\01-Phan-loai\KNN-Classifier\diabetes.csv')
df.info()
df.head()
df.shape
```

```python
x = df.drop(['Outcome'],axis = 1)
y = df['Outcome']

COT_KHONG_DUOC_PHEP_0 = ['Glucose','BloodPressure','SkinThickness','Insulin','BMI']

x[COT_KHONG_DUOC_PHEP_0] = x[COT_KHONG_DUOC_PHEP_0].replace(0, np.nan)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.impute import SimpleImputer
from sklearn.ensemble import BaggingClassifier

pipeline = Pipeline([
    ('imp',SimpleImputer(strategy='mean')),
    ('scale',StandardScaler()),
    ('knn', KNeighborsClassifier(n_neighbors = 4)),
])

```

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.metrics import accuracy_score

grid = {
    'knn__n_neighbors': list(range(1,32,2)),
    'knn__weights' : ['uniform', 'distance'],
    'knn__metric' : ['euclidean','manhattan']
}

gs = GridSearchCV(pipeline,grid, cv = StratifiedKFold(5, shuffle=True,random_state=42),scoring='recall',n_jobs = -1)
gs.fit(x,y)




print(gs.best_params_)
```
